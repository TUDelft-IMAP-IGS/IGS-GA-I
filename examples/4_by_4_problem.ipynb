```python
stakeholder_preferences = {
     1: {
        "1_profit": lambda p: p1(p),
        "2_noise": lambda n: p2(n),
        "3_mortality": lambda m: p3(m),
        "4_pollution": lambda po: p4(po),
        },
    2: {
        "1_profit": lambda p: p5(p),
        "2_noise": lambda n: p6(n),
        "3_mortality": lambda m: p7(m),
        "4_pollution": lambda po: p8(po),
        },
    3: {
        "1_profit": lambda p: p9(p),
        "2_noise": lambda n: p10(n),
        "3_mortality": lambda m: p11(m),
        "4_pollution": lambda po: p12(po),
        },
    4: {
        "1_profit": lambda p: p13(p),
        "2_noise": lambda n: p14(n),
        "3_mortality": lambda m: p15(m),
        "4_pollution": lambda po: p16(po),
        },
}
```

```python
objective_weights = {
    1: {"1_profit": 1, "2_noise": 0, "3_mortality": 0, "4_pollution": 0},
    2: {"1_profit": 0.3, "2_noise": 0.4, "3_mortality": 0, "4_pollution": 0.3},
    3: {"1_profit": 0.4, "2_noise": 0, "3_mortality": 0.4, "4_pollution": 0.2},
    4: {"1_profit": 0.3, "2_noise": 0.1, "3_mortality": 0.1, "4_pollution": 0.5},
}
```

```python
# Set optimization options
optimization_options = {
    'population_size': 200,
    'generations': 15,
    'mutation_rate': 0.8,
    'hall_of_fame_size': 200,
}
```

```python
@Preferendus(
     preference_functions=stakeholder_preferences,
     stakeholder_weights=[0.25, 0.25, 0.25, 0.25],
     variable_types=[float, int, int, float],
     value_range=[(2.0, 10.0), (1.0, 12.0), (50, 150), (3, 15)],
     objective_weights=objective_weights,
     optimization_options=optimization_options,
)
def run_simulation(x):
    # Simulation logic here
    F1 = objectives.objective_profit(x)
    F2 = objectives.objective_noise(x)
    F3 = objectives.objective_mortality(x)
    F4 = objectives.objective_particle(x)
    return None, None, F1, F2, F3, F4
```

```python
result = run_simulation([15,20,150,10])  # Initial guess or can be omitted
print(f"Optimal variables: {result['x_opt']}, Utility: {result['utility']}")
```
